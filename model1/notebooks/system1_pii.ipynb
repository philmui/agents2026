## Create a TypeSafeClassifier

```python
from langchain_typesafe import Noul, TypeSafeClassifier
```

```python
classifier = TypeSafeClassifier()
```

```python
response = classifier.invoke({
    "state": (
        "The document is a legal contract between 2 parties. "
        "It outlines the terms and conditions of the agreement, including payment details, "
        "obligations of each party, and the duration of the contract. "
        "Can you tell me if it has personally identifiable information (PII) in it?"
    ),
    "questions": {
        "pii": Noul(
            instructions="Does the document contain personally identifiable information (PII)?"),
    },
})
```

```python
pii = response.nouls["pii"]
type(pii), pii
```
